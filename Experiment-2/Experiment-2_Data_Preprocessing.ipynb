```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler

print("Libraries imported successfully!")
```

```python

# Create Food Delivery dataset

data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005,
                 1006, 1002, 1008, 1009, 1010],

    "Customer": ["Aarav", "Diya", "Kabir", "Meera", "Ishaan",
                 "Anaya", "Diya", "Rohan", "Sara", "Vivaan"],

    "Food_Category": ["Pizza", "Indian", "Chinese", "Burgers", None,
                      "Indian", "Indian", "Mexican", "Pizza", "Chinese"],

    "City": ["Mumbai", "Thane", "Mumbai", "Navi Mumbai", "Pune",
             "Thane", "Thane", "Mumbai", "Pune", "Navi Mumbai"],

    "Delivery_Time": [30, 40, np.nan, 25, 35, 50, 40, 20, 45, 55],

    "Order_Amount": [450, 300, 550, 250, 400,
                     350, 300, 600, 500, 700],

    "Rating": [4.5, 4.0, 5.0, np.nan, 4.2,
               3.8, 4.0, 4.7, 4.4, 4.1],

    "Payment_Mode": ["UPI", "Card", "Cash", "UPI", "Card",
                     None, "Card", "UPI", "Cash", "Card"]
}

df = pd.DataFrame(data)

print("===== ORIGINAL FOOD DELIVERY DATASET =====")
print(df)

print("\nDataset Shape:", df.shape)
```

```python

# Identify missing values

print("===== MISSING VALUES IN FOOD DELIVERY DATASET =====")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

```

```python

# Handle missing values

# Fill numerical missing values using mean
df["Delivery_Time"] = df["Delivery_Time"].fillna(
    df["Delivery_Time"].mean()
)

df["Rating"] = df["Rating"].fillna(
    df["Rating"].mean()
)

# Fill categorical missing values using mode
df["Food_Category"] = df["Food_Category"].fillna(
    df["Food_Category"].mode()[0]
)

df["Payment_Mode"] = df["Payment_Mode"].fillna(
    df["Payment_Mode"].mode()[0]
)

print("===== DATASET AFTER HANDLING MISSING VALUES =====")
print(df)

print("\n===== MISSING VALUES AFTER HANDLING =====")
print(df.isnull().sum())
```

```python

# Remove duplicate records

df = df.drop_duplicates()

print("===== DATASET AFTER REMOVING DUPLICATES =====")
print(df)

print("\nDuplicates after removal:", df.duplicated().sum())
print("Dataset Shape:", df.shape)
```

```python

# Encode categorical columns

food_encoder = LabelEncoder()
city_encoder = LabelEncoder()
payment_encoder = LabelEncoder()

df["Food_Category"] = food_encoder.fit_transform(df["Food_Category"])

df["City"] = city_encoder.fit_transform(df["City"])

df["Payment_Mode"] = payment_encoder.fit_transform(df["Payment_Mode"])

print("===== DATASET AFTER CATEGORICAL ENCODING =====")
print(df)

# Display encoding mappings
print("\nFood Category Encoding:")
print(dict(zip(food_encoder.classes_,
               food_encoder.transform(food_encoder.classes_))))

print("\nCity Encoding:")
print(dict(zip(city_encoder.classes_,
               city_encoder.transform(city_encoder.classes_))))

print("\nPayment Mode Encoding:")
print(dict(zip(payment_encoder.classes_,
               payment_encoder.transform(payment_encoder.classes_))))
```

```python

# Feature Scaling using StandardScaler

numerical_columns = [
    "Delivery_Time",
    "Order_Amount",
    "Rating"
]

scaler = StandardScaler()

df[numerical_columns] = scaler.fit_transform(
    df[numerical_columns]
)

print("===== DATASET AFTER FEATURE SCALING =====")
print(df)

print("\n===== SCALED NUMERICAL COLUMNS =====")
print(df[numerical_columns])
```

```python

# Final dataset verification

print("===== FINAL DATASET VERIFICATION =====")

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nFinal Dataset Shape:", df.shape)

print("\nFinal Preprocessed Dataset:")
print(df)
```

```python

# Export preprocessed dataset to Excel

output_file = "Preprocessed_Food_Delivery.xlsx"

df.to_excel(output_file, index=False)

print("Preprocessed dataset saved successfully!")
print("File name:", output_file)
```
